## 3. Una gráfica

Graficamos un oscilador armónico amortiguado:

$$y(t) = e^{-t/4} \cos(2\pi t)$$

Si ves la curva abajo, Plotly renderiza correctamente dentro de VS Code.

```python
import plotly.graph_objects as go

t = np.linspace(0, 10, 400)
y = np.exp(-t / 4) * np.cos(2 * np.pi * t)

fig = go.Figure()
fig.add_trace(go.Scatter(x=t, y=y, mode="lines", line=dict(width=2)))
fig.update_layout(
    title="Oscilador amortiguado",
    xaxis_title="t [s]",
    yaxis_title="amplitud",
    template="plotly_white",
)
fig.show()
```
